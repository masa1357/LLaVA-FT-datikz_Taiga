```python
from sklearn.model_selection import GroupKFold
from torch.utils.data import Subset, ConcatDataset, DataLoader
```

```python
import src.plotlylayout as pl
import plotly.graph_objects as go
pl.dark_mode()

def plot_fold_split_plotly(n_samples, train_idx, valid_idx, title=None):
    import plotly.graph_objects as go

    fig = go.Figure().update_layout(
        margin=dict(l=10, r=10, t=20, b=50)
    )
    fig.add_trace(go.Scatter(
        x=train_idx, y=[0]*len(train_idx),
        mode="markers",
        marker=dict(symbol="line-ns-open", size=18),
        name="train"
    ))
    fig.add_trace(go.Scatter(
        x=valid_idx, y=[0]*len(valid_idx),
        mode="markers",
        marker=dict(symbol="line-ns-open", size=22),
        name="valid"
    ))
    fig.update_layout(
        title=title,
        xaxis_title="sample index",
        yaxis=dict(showticklabels=False, showgrid=False, zeroline=False),
        height=200,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
    )
    return fig
```

```python
# tokenizerのロード
from transformers import AutoTokenizer

base_model = "elyza/Llama-3-ELYZA-JP-8B"
tokenizer = AutoTokenizer.from_pretrained(
    base_model, use_fast=True  # padding_side="left"
)
tokenizer.padding_side = "left"
if tokenizer.pad_token is None:
    tokenizer.add_special_tokens({"pad_token": "[PAD]"})
```

```python
ans = "今日はいい天気！"
tokens = tokenizer.encode(ans, add_special_tokens=False, truncation=False)
print(tokens)
dec = tokenizer.decode(tokens, skip_special_tokens=True)
print(dec)
```
