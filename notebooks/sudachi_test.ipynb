```python
# sudachipyの動作を確認
def get_wakati(dataset, question_filter = [1,2,3,4,5]):
    dict = sudachipy.Dictionary()
    tokenizer = dict.create() # 辞書から分割器を作る

    ret_data = []

    # 単語分かち書き
    for sample in dataset:

        # 新しい sample データを作る
        sample_wakati = {
            "userid": sample["userid"],
            "labels": sample["labels"],
            "grades": sample["grades"],
            "setences": sample["sentences"],  # 元の回答を保持
            "wakati": {}  # 分かち書き用の新しい階層
        }

        for c in range(1, 16):
            c_key = f"L{c}"
            sample_wakati["wakati"][c_key] = {}

            for q in question_filter:
                q_key = f"Q{q}"
                original_text = sample["sentences"][c_key][q_key]

                # 分かち書き実行
                wakati = tokenizer.tokenize(original_text)
                sample_wakati["wakati"][c_key][q_key] = [m.surface() for m in wakati]

        ret_data.append(sample_wakati)

    return ret_data
```

```python
from gensim.models import Word2Vec
from scipy.spatial.distance import cosine
```

```python
class Word2VecModel:
    def __init__(self, train_data:List[Dict[str, object]], question_filter:List[int] = [1,2,3,4,5]):
        
        self.LABEL_MAPPING = {"A": 0, "B": 1, "C": 2, "D": 3, "F": 4}
        MODEL_PATH = DATA_PATH / "word2vec"
        
        # word2vec parameters
        self.num_features = 200
        self.min_word_count = 5
        self.num_workers = 40
        self.context = 10
        self.downsampling = 1e-3
        self.model_name = "word2vec_testcase.model"
        self.MODEL_PATH = MODEL_PATH/self.model_name
        # コーパス読み込み
        self.corpus = [token
            for sample in train_data
            for c in range(1, 16)
            for q in question_filter
            for token in sample["wakati"][f"L{c}"][f"Q{q}"]]
        

        self.model = self.get_model()

    def get_model(self):
        if self.MODEL_PATH.exists():
            print(f"Loading model from {self.MODEL_PATH}")
            self.model = Word2Vec.load(str(self.MODEL_PATH))
        else:
            print("Training new Word2Vec model")
            model = Word2Vec(
                self.corpus,
                workers=self.num_workers,
                hs=1,
                sg=1,
                negative=10,
                epochs=25,
                vector_size=self.num_features,
                min_count=self.min_word_count,
                window=self.context,
                sample=self.downsampling,
                seed=1,
            )
            # self.model.save(str(self.MODEL_PATH))
            # self.logger.info(f"saved word2vec model:{self.MODEL_PATH}")
        print(f"Word2Vec model loaded with {len(model.wv.index_to_key)} words.")

        return model

    def wordvec2docvec(self, sentence):
        # 文章ベクトルの初期値（0ベクトルを初期値とする）
        docvecs = np.zeros(self.num_features, dtype="float32")

        # 文章に現れる単語のうち、モデルに存在しない単語をカウントする
        denomenator = len(sentence)

        # 文章内の各単語ベクトルを足し合わせる
        for word in sentence:
            try:
                temp = self.model.wv[word]
                # docvecs += temp
            except:
                denomenator -= 1
                # print(f"{word}はモデルに存在しません。")
                continue
            docvecs += temp

        # 文章に現れる単語のうち、モデルに存在した単語の数で割る
        if denomenator > 0:
            docvecs = docvecs / denomenator

        return docvecs
    
    def get_vector(self, dataset, question_filter=[1, 2, 3, 4, 5]):
        ret_data = []
        for sample in dataset:
            sample_vector = {
                "userid": sample["userid"],
                "labels": sample["labels"],
                "grades": sample["grades"],
                "vector": {},  # ベクトル化用の新しい階層
            }

            for c in range(1, 15):
                c_key = f"L{c}"
                next_key = f"L{c+1}"
                vec_key= f"L{c}_{c+1}"
                sample_vector["vector"][vec_key] = {}

                for q in question_filter:
                    q_key = f"Q{q}"               

                    # ベクトル化
                    vector_1 = self.wordvec2docvec(sample["wakati"][c_key][q_key])
                    vector_2 = self.wordvec2docvec(sample["wakati"][next_key][q_key])
                    
                    # ベクトル間距離を計算
                    # ユークリッド距離
                    distance = np.linalg.norm(vector_1 - vector_2)
                    # コサイン距離（ユークリッド距離と同様に，値が大きいほど意味的に離れている）
                    if np.linalg.norm(vector_1) == 0 or np.linalg.norm(vector_2) == 0:
                        return 0.0
                    cosine_distance = 1 - cosine(vector_1, vector_2)

                    # 結果を格納
                    sample_vector["vector"][vec_key][q_key] = {
                        "euc_dist": distance,
                        "cos_dist": cosine_distance,
                    }
                    
            ret_data.append(sample_vector)
        return ret_data
```
